## Section V: Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df_clean= pd.read_csv("Emesis_Norovirus_Clean.csv")
```

### Visualization of Volume Using Correlation Matrix

```python
plt.figure(figsize=(10,6)) 
sns.heatmap(df_clean.corr(), annot=True, cmap='coolwarm')
plt.title("Correlations Between Variables")
plt.savefig('correlationmatrix2.jpg')
plt.show()
```

Volume has a negative correlation with every numerical data. Therefore, when volume increases, the pH and titer decrease. Therefore, when the pH of the vomit is lower (more acidic), a patient is more likely to produce more vomit. When the volume of the vomit is large, then the titer of the virus is usually decreased. This makes sense since an increasing volume would lower the concentration of any substance.
The titer and pH have a positive correlation to each other. This proves my claim from earlier that the titer and pH have a positive/symbiotic relationship with each other. All of these correlations are relatively low, and not close to 0.5 let alone 1.0. 

## Finding the Optimal pH by Creating a Line Plot

```python
mean_titer= df_clean.groupby('pH')['titer'].mean()
plt.plot(mean_titer, marker='o')
plt.xlabel('pH')
plt.ylabel('Mean Titer')
plt.title('Optimal pH for Titer')
plt.savefig("linegraph1.jpg")
plt.show()
```

Here, we can see that there is an optimal pH value that results in the highest amount of titer. The optimal pH is between the range of 5-6. This is an interesting finding that indicates that the norovirus thrives in vomit, in slightly acidic environments instead of extremely acidic/basic conditions.
